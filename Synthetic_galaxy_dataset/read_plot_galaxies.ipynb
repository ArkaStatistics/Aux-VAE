### Image parameters :

    params : Array with :
    - flux : Flux of the galaxy in counts.
    - radius : Radius of the galaxy in arcsec.
    - g1 : Reduced shear component.
    - g2 : Reduced shear component. See demo2 comments. sqrt(g1**2+g2**2) < 1.0.
    - psf_fwhm : fwhm of the Gaussian PSF used for convolution.

```python
dirIn = './data/'

fileIn = dirIn + 'lhc_16384_5_training' + '.txt'  ## Training set N=256


# Read image params from the text file
image_params = catalog_read_plot_utils.read_input_params(fileIn)
print('Parameter shape (nGal, param_size):', image_params.shape)
```

```python
# Parameter distribution of part of the data
column_labels = ['flux', 'radius', 'g1', 'g2', 'psf']
catalog_read_plot_utils.plot_scatter_and_histogram_labeled(image_params[np.random.randint(0, image_params.shape[0], size=1000)], column_labels)
```
